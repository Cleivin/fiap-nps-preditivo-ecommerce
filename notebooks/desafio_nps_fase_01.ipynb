# 01 - Entendendo o Negócio

## 1) Qual problema de negócio está sendo resolvido?

O principal problema é que o e-commerce acaba percebendo a insatisfação do cliente tarde demais. Muitas vezes, isso só fica claro depois que o pedido foi entregue e o cliente respondeu à pesquisa de NPS, ou simplesmente quando ele deixa de comprar.

Enquanto isso, problemas como atraso na entrega, frete, atendimento do SAC, problemas com o pedido ou outras situações já podem ter prejudicado a experiência. Quando a empresa percebe, muitas vezes já não há muito o que fazer.

A ideia é justamente entender o que pode estar levando um cliente a ter uma experiência ruim e identificar esses sinais antes que isso aconteça. Assim, a empresa pode agir de forma mais preventiva para reduzir o risco de perder clientes, diminuir a recompra ou gerar uma experiência negativa.

## 2) Por que o NPS é importante para um e-commerce?

O NPS é importante porque ajuda a empresa a entender, de forma simples, como o cliente percebeu a experiência que teve com a marca.

- **É um indicador de satisfação:** ajuda a acompanhar se os clientes estão tendo uma experiência positiva ou negativa com a empresa.

- **Transforma a percepção do cliente em um número:** as notas permitem separar os clientes entre promotores, neutros e detratores, facilitando o acompanhamento da satisfação ao longo do tempo.

- **Pode indicar problemas na operação:** uma queda no NPS pode ser um sinal de que algo está acontecendo na experiência do cliente, como problemas de entrega, atendimento, prazo ou pedido.

- **Tem relação com a recompra:** um cliente que teve uma experiência ruim pode ter menos interesse em comprar novamente. Por isso, acompanhar o NPS pode ajudar a identificar riscos de perda de clientes.

- **Ajuda na tomada de decisão:** olhando o NPS junto com os dados dos pedidos, a empresa consegue entender melhor o que está influenciando a experiência do cliente e onde precisa atuar.

**Em resumo:** o NPS funciona como um **termômetro da experiência do cliente**. Ele não explica sozinho por que o cliente ficou satisfeito ou insatisfeito, mas ajuda a identificar quando existe um problema e onde vale a pena investigar.

## 3) Quais áreas poderiam se beneficiar desses insights?

Os insights gerados podem ser úteis para várias áreas da empresa, principalmente aquelas que têm contato direto com a experiência do cliente.

- **Logística / Supply:** entender quais transportadoras, regiões, rotas ou tipos de entrega estão mais relacionados a clientes insatisfeitos. Com isso, a empresa pode agir antes que o problema aconteça, por exemplo, entrando em contato com o cliente em caso de atraso ou revisando prazos.

- **Atendimento (SAC / CX):** identificar pedidos com maior risco de gerar insatisfação e permitir que a equipe se antecipe às reclamações. Isso pode ajudar a reduzir chamados e melhorar a experiência do cliente.

- **Produto / Sortimento:** analisar se determinados produtos, categorias ou combinações de itens aparecem com mais frequência entre clientes insatisfeitos. Esses dados podem ajudar a ajustar a oferta ou até mesmo a forma como o produto é apresentado no site.

- **Pricing / Promoções:** entender se descontos e promoções realmente contribuem para uma experiência melhor. Um desconto pode atrair o cliente, mas se o pedido atrasar ou houver algum problema no atendimento, o efeito positivo da promoção pode acabar sendo perdido.

- **Marketing / CRM:** usar os insights para separar clientes com diferentes níveis de satisfação e criar ações mais específicas, como campanhas de retenção para clientes em risco ou ações de relacionamento com clientes mais satisfeitos.

- **Financeiro / Diretoria:** relacionar a satisfação dos clientes com indicadores como recompra, custos de atendimento e possíveis perdas de receita. Isso ajuda a entender onde os investimentos podem gerar maior impacto.

- **Dados / IA:** investigar os dados, identificar padrões e testar hipóteses para entender quais fatores estão mais relacionados à insatisfação. Em uma etapa posterior, esses insights também podem servir como base para um modelo preditivo.

No fim, o objetivo não é apenas descobrir quais fatores estão relacionados ao NPS, mas transformar essas informações em ações. O insight só gera valor quando alguma área consegue utilizá-lo para tomar uma decisão diferente.

## 4) Reflexão — como o NPS impacta recompra, boca a boca e market share

O NPS pode ter impacto em vários pontos do negócio, principalmente porque a experiência que o cliente teve influencia não só a percepção sobre a empresa, mas também o que ele faz depois da compra.

**Recompra**

- Clientes que tiveram uma experiência positiva tendem a ter mais motivos para voltar a comprar, enquanto uma experiência ruim pode diminuir a intenção de comprar novamente.

- Na nossa base existe a informação `repeat_purchase_30d`, que pode ser usada na análise para verificar se existe alguma relação entre a nota do NPS e a recompra.

- Porém, essa variável precisa de atenção caso seja usada em um modelo preditivo. Se a previsão for feita no momento da entrega, por exemplo, ainda não saberíamos se o cliente comprou novamente nos 30 dias seguintes. Nesse caso, usar essa informação no modelo seria um possível caso de leakage.

**Boca a boca**

- O próprio NPS está relacionado à intenção de recomendar a empresa. Por isso, clientes satisfeitos podem contribuir para uma percepção positiva da marca.

- Por outro lado, clientes insatisfeitos podem deixar avaliações negativas, fazer reclamações ou compartilhar uma experiência ruim nas redes sociais.

- No e-commerce, isso pode ter um impacto ainda maior porque uma experiência negativa pode ser compartilhada rapidamente e alcançar muitas outras pessoas, por exemplo ter muitas reclamações do site Reclame Aqui pode impactar negativamente para empresa.

**Market share**

- O crescimento de participação de mercado depende tanto de conquistar novos clientes quanto de conseguir manter os atuais.

- Uma experiência ruim pode aumentar a chance de o cliente procurar outra empresa, além de dificultar a aquisição de novos clientes quando a reputação da marca é afetada.

- Já uma experiência positiva pode contribuir para retenção e indicação, tornando o crescimento mais sustentável.

**Cadeia lógica**

Uma possível relação que queremos investigar é:

`Problemas na operação >> experiência ruim >> NPS mais baixo >> menos recompra e mais reclamações >> impacto na retenção e aquisição de clientes`

## 5) Quais indicadores de mercado poderiam complementar a análise?

Além dos dados que já temos, alguns indicadores externos poderiam ajudar a dar mais contexto para os resultados. A ideia seria entender se determinados problemas são específicos da empresa ou se também aparecem no mercado como um todo.

- **Benchmark de NPS do setor:** permitiria comparar o NPS da empresa com outras empresas de e-commerce e entender se o resultado está dentro de um nível esperado para o setor.

- **SLA logístico:** informações sobre o percentual de entregas realizadas no prazo e o atraso médio do mercado ajudariam a entender se os problemas de entrega são específicos da operação ou se estão relacionados a dificuldades comuns de determinada região ou transportadora.

- **OTIF (On Time In Full):** além de verificar se o pedido chegou no prazo, esse indicador considera se ele chegou completo e corretamente. Isso poderia ajudar a entender melhor a relação entre problemas na entrega e satisfação do cliente.

- **CSAT / CES:** outras métricas de satisfação poderiam complementar o NPS. O CSAT, por exemplo, pode ajudar a avaliar uma experiência específica, enquanto o CES mede o esforço que o cliente teve para resolver algum problema. Temos uma CSAT por atendimento na base de dados.

- **Taxa de recompra e retenção:** ajudariam a entender se clientes mais satisfeitos realmente voltam a comprar. Na nossa base, o `repeat_purchase_30d` pode ser utilizado para começar essa análise.

- **CAC (Custo de Aquisição de Cliente) e LTV (Valor do Tempo de Vida do Cliente):** esses indicadores ajudariam a conectar a satisfação do cliente com o impacto financeiro. Se clientes satisfeitos compram mais vezes ou permanecem mais tempo na empresa, isso pode ter impacto direto no valor gerado por cada cliente.

- **Reviews e presença da marca:** avaliações em sites, marketplaces e redes sociais poderiam complementar o NPS mostrando como os clientes estão falando da empresa fora da pesquisa.

- **Indicadores do SAC:** métricas como quantidade de contatos, tempo médio de atendimento, tempo para resolução e número de recontatos ajudariam a entender se problemas no atendimento estão relacionados à insatisfação.

- **Ticket médio e promoções:** comparar o valor dos pedidos e os descontos aplicados com os resultados de satisfação pode ajudar a entender se preço e promoções influenciam a experiência do cliente.

Esses indicadores não precisam necessariamente estar disponíveis agora. Eles podem ser usados posteriormente para complementar a análise e dar mais contexto aos resultados encontrados na nossa base.

# 02 - Definição de Target

## 1) Qual variável representa a satisfação do cliente?

A principal variável que representa a satisfação do cliente na nossa base é a `nps_score`, que contém a nota dada pelo cliente, de 0 a 10, em relação à experiência que teve com aquele pedido.

A partir dessa nota, também podemos criar uma classificação do cliente:

- **0 a 6 — Detrator:** cliente que teve uma experiência ruim ou ficou insatisfeito.
- **7 a 8 — Neutro:** cliente que teve uma experiência razoável, mas não necessariamente ficou satisfeito a ponto de recomendar.
- **9 a 10 — Promotor:** cliente que teve uma experiência positiva e tem maior probabilidade de recomendar a empresa.

Para a análise, podemos trabalhar com a `nps_score` de duas formas:

- **Nota do NPS:** analisar diretamente a nota de 0 a 10. Isso permite entender quais fatores estão relacionados a notas maiores ou menores.

- **Classificação do NPS:** separar os clientes em detratores, neutros e promotores. Essa abordagem pode ser mais interessante para o negócio, principalmente quando o foco é identificar clientes que tiveram uma experiência ruim.

É importante não confundir a `nps_score` com outras variáveis da base. A `csat_internal_score` representa o **CSAT (Índice de Satisfação do Cliente)** e é uma outra forma de medir satisfação, porém de um atendimento específico, por exemplo, cliente que ligou no SAC para tirar um dúvida.

## 2) Por que ela foi escolhida?

A `nps_score` foi escolhida porque é a variável que melhor representa o problema que queremos entender: a satisfação do cliente depois da experiência de compra.

- **Está diretamente ligada ao problema do case:** o foco é entender a insatisfação dos clientes e identificar possíveis causas de uma experiência ruim. O NPS é uma forma direta de medir isso.

- **É um indicador relevante para o negócio:** acompanhar a satisfação dos clientes ajuda a empresa a entender se a experiência oferecida está de acordo com o esperado.

- **Está disponível na nossa base:** a informação está no CSV por meio da variável `nps_score`, o que permite analisar tanto a nota individual quanto criar as categorias de detrator, neutro e promotor.

- **Pode gerar ações:** identificar uma concentração de detratores pode indicar a necessidade de investigar problemas na operação, enquanto uma base maior de promotores pode representar uma oportunidade de retenção e indicação.

- **Permite diferentes tipos de análise:** podemos analisar a nota de 0 a 10 ou transformar o NPS em categorias. Para o negócio, identificar principalmente os detratores pode ser interessante porque são os clientes que tiveram uma experiência mais negativa.

- **É uma métrica padronizada:** apesar de ser coletada no final da experiência, o NPS continua sendo uma forma simples e padronizada de medir a satisfação dos clientes.

**Em resumo:** escolhemos a `nps_score` porque ela representa diretamente a satisfação que queremos analisar, está disponível nos dados e permite transformar o resultado em informações que podem ser utilizadas pelo negócio.

#### Um ponto importante sobre antecipação

O NPS só é conhecido depois que o cliente responde à pesquisa. Por isso, ele será nosso **target**.

Para antecipar uma possível insatisfação, precisamos usar informações que já estejam disponíveis durante a jornada, como contatos com o SAC, atrasos, reclamações e dados de logística.

## 3) Em que momento da jornada essa informação é coletada?

Atualmente, o NPS é coletado apenas após o encerramento da jornada de compra. Isso limita a capacidade da empresa de identificar problemas com antecedência, priorizar ações corretivas e atuar de forma preventiva.

Nesse contexto, surge a necessidade de utilizar dados operacionais, como informações de pedidos, logística e atendimento, para identificar possíveis sinais de insatisfação e transformá-los em insights que apoiem decisões estratégicas e operacionais.

## 4) Existe algum risco de usar essa variável de forma inadequada?

**Sim.** Apesar de o `nps_score` ser uma variável importante para o case, existem alguns cuidados para evitar interpretações ou análises incorretas.

- **Usar o NPS como se fosse a satisfação completa:** o NPS é uma forma de medir a percepção do cliente, mas não representa necessariamente toda a experiência ou lealdade à marca. Por isso, devemos deixar claro que estamos analisando a **nota de NPS** e não a satisfação absoluta do cliente.

- **Viés de amostragem:** nem todos os clientes necessariamente respondem à pesquisa de NPS. Isso significa que os clientes presentes na base de respostas podem não representar todos os clientes da empresa.

- **Diferenças entre grupos:** o comportamento do NPS pode variar entre diferentes grupos de clientes, como regiões ou faixas etárias. É importante verificar se existem diferenças relevantes antes de tirar conclusões gerais.

- **Confundir pedido com cliente:** neste case, o NPS está relacionado à experiência de um **pedido**. Portanto, a unidade de análise deve ser o `order_id`, e não necessariamente o cliente ou a marca como um todo. O mesmo cliente pode ter diferentes avaliações conforme o pedido.

#### Legenda dos tipos

| Tipo | O que significa | Exemplo na base |
|------|-----------------|-----------------|
| **ID** | Código que **identifica** uma entidade. Não se soma, não se tira média, não entra como “número” no modelo. | `customer_id`, `order_id` |
| **Contínua** | Número que pode ter **casas decimais** (ou muitos valores distintos em uma escala). Mede quantidade/intensidade. | `order_value`, `nps_score`, `freight_value` |
| **Discreta** | Número **inteiro / contagem** (0, 1, 2, 3…). Ainda é quantitativa, mas em saltos. | `items_quantity`, `customer_age`, `complaints_count` |
| **Nominal** | **Categoria** sem ordem (rótulo). Não faz sentido dizer que um valor é “maior” que outro. | `customer_region` |
| **Ordinal** | Categoria **com ordem** (baixo < médio < alto), mas os intervalos não são necessariamente iguais. | *(não usamos nesta tabela; ex.: faixa de satisfação)* |
| **Booleana / binária** | Só **dois estados** (sim/não, 0/1). | `repeat_purchase_30d` |
| **Target** | Variável que queremos **explicar ou prever** — no case, a satisfação. | `nps_score` |

Em resumo: **qualitativa** = nominal / ordinal / booleana; **quantitativa** = discreta / contínua; **ID** fica de fora da análise numérica.

## Parte 3 — Descritivas e cortes de negócio

**Objetivo:** entender a distribuição do NPS e como ele muda por atraso, SAC, região e perfil — ainda sem teste formal.

Antes de tudo: aplicar o **filtro do veredito (2.6)** em `df_mov_ecom_bruto` e seguir a análise em `df_mov_ecom_filtrado`.

### 3.0 — Aplicar o filtro do veredito

Excluímos:
1. resolução > 0 e/ou CSAT > 0 **sem** SAC e **sem** reclamação;
2. atraso > prazo total de entrega.

O bruto permanece em `df_mov_ecom_bruto`; a análise daqui pra frente usa `df_mov_ecom_filtrado`.

```python
sem_atendimento = (
    (df_mov_ecom_bruto["customer_service_contacts"] == 0)
    & (df_mov_ecom_bruto["complaints_count"] == 0)
)
lixo_atendimento = sem_atendimento & (
    (df_mov_ecom_bruto["resolution_time_days"] > 0)
    | (df_mov_ecom_bruto["csat_internal_score"] > 0)
)
atraso_incoerente = df_mov_ecom_bruto["delivery_delay_days"] > df_mov_ecom_bruto["delivery_time_days"]

mask_excluir = lixo_atendimento | atraso_incoerente
df_mov_ecom_filtrado = df_mov_ecom_bruto.loc[~mask_excluir].copy()

print(f"Antes (df_mov_ecom_bruto): {len(df_mov_ecom_bruto)}")
print(f"Excluídos — atendimento inconsistente: {lixo_atendimento.sum()}")
print(f"Excluídos — atraso > prazo: {atraso_incoerente.sum()}")
print(f"Excluídos — total: {mask_excluir.sum()} ({mask_excluir.mean():.1%})")
print(f"Depois (df_mov_ecom_filtrado): {len(df_mov_ecom_filtrado)}")
```

**Insight (3.0):** filtro aplicado conforme veredito 2.6 — saíram ~**144** pedidos (~**5,8%**): ~23 com resolução/CSAT sem SAC e sem reclamação + ~121 com atraso > prazo. Restam ~**2.356** pedidos em `df_mov_ecom_filtrado` para a EDA.

### 3.1 — Colunas auxiliares de negócio

Criamos classes NPS (Reichheld), flags e faixas para os `groupby` seguintes. Com NPS em float, os cortes usam faixas inclusivas.

```python
# Classes clássicas NPS (Reichheld). Com float, cortes inclusivos:
df_mov_ecom_filtrado["nps_classe"] = pd.cut(
    df_mov_ecom_filtrado["nps_score"],
    bins=[-0.01, 6, 8, 10],
    labels=["detrator", "neutro", "promotor"],
)

df_mov_ecom_filtrado["eh_detrator"] = (df_mov_ecom_filtrado["nps_score"] <= 6).astype(int)
df_mov_ecom_filtrado["tem_atraso"] = (df_mov_ecom_filtrado["delivery_delay_days"] > 0).astype(int)

df_mov_ecom_filtrado["faixa_atraso"] = pd.cut(
    df_mov_ecom_filtrado["delivery_delay_days"],
    bins=[-0.01, 0, 2, 4, 8],
    labels=["0 dias", "1-2 dias", "3-4 dias", "5+ dias"],
)

df_mov_ecom_filtrado["faixa_sac"] = pd.cut(
    df_mov_ecom_filtrado["customer_service_contacts"],
    bins=[-0.01, 0, 1, 3, 10],
    labels=["0", "1", "2-3", "4+"],
)

df_mov_ecom_filtrado["faixa_idade"] = pd.cut(
    df_mov_ecom_filtrado["customer_age"],
    bins=[17, 29, 44, 59, 100],
    labels=["18-29", "30-44", "45-59", "60+"],
)

df_mov_ecom_filtrado["faixa_tenure"] = pd.cut(
    df_mov_ecom_filtrado["customer_tenure_months"],
    bins=[0, 12, 36, 60, 200],
    labels=["ate_12m", "13-36m", "37-60m", "60m+"],
)

df_mov_ecom_filtrado["faixa_ticket"] = pd.qcut(
    df_mov_ecom_filtrado["order_value"], q=4, labels=["Q1_baixo", "Q2", "Q3", "Q4_alto"]
)

df_mov_ecom_filtrado["faixa_reclamacoes"] = pd.cut(
    df_mov_ecom_filtrado["complaints_count"],
    bins=[-0.01, 2, 5, 11],
    labels=["0-2", "3-5", "6+"],
)

print("Colunas auxiliares criadas.")
df_mov_ecom_filtrado[["nps_score", "nps_classe", "eh_detrator", "faixa_atraso", "faixa_sac"]].head()
```

**Insight (3.1):** colunas auxiliares criadas — `nps_classe`, `eh_detrator`, `tem_atraso` e faixas (atraso, SAC, idade, tenure, ticket, reclamações). Prontas para os `groupby` e gráficos da Parte 3/4.

Distribuição das classes NPS e o NPS da amostra (% promotores − % detratores).

```python
nps_dist = (
    df_mov_ecom_filtrado["nps_classe"]
    .value_counts()
    .rename_axis("nps_classe")
    .reset_index(name="qtd")
)
nps_dist["%share"] = (nps_dist["qtd"] / nps_dist["qtd"].sum() * 100).round(1)

pct_promotor = (df_mov_ecom_filtrado["nps_classe"] == "promotor").mean()
pct_detrator = (df_mov_ecom_filtrado["nps_classe"] == "detrator").mean()
nps_amostra = pct_promotor - pct_detrator

total = pd.DataFrame(
    [{"nps_classe": "Total", "qtd": nps_dist["qtd"].sum(), "%share": 100.0}]
)
display(pd.concat([nps_dist, total], ignore_index=True))
print(f"% promotores: {pct_promotor:.1%}")
print(f"% detratores: {pct_detrator:.1%}")
print(f"NPS da amostra (% promotores - % detratores): {nps_amostra:.1%}")
```

**Insight (3.1 — classes NPS):** na base filtrada, ~**73,5%** detratores (1.732), ~**18,5%** neutros (437) e ~**7,9%** promotores (187). NPS da amostra (% promotores − % detratores) ≈ **−65,6 p.p.** — ainda uma base muito insatisfeita, mesmo após limpar o lixo lógico.

### 3.2 — Tendência central e dispersão do NPS

Além do `describe`: média, mediana, moda, desvio, CV, quartis e IQR (critério clássico de outlier).

```python
s = df_mov_ecom_filtrado["nps_score"]

print("média:", round(s.mean(), 3))
print("mediana:", round(s.median(), 3))
print("moda:", s.mode().iloc[0])
print("desvio padrão:", round(s.std(), 3))
print("variância:", round(s.var(), 3))
print("CV (std/média):", round(s.std() / s.mean(), 3))

q1, q2, q3 = s.quantile([0.25, 0.5, 0.75])
iqr = q3 - q1
print("Q1, Q2, Q3:", round(q1, 3), round(q2, 3), round(q3, 3))
print("IQR:", round(iqr, 3))
print("limite inferior típico (outliers):", round(q1 - 1.5 * iqr, 3))
print("limite superior típico (outliers):", round(q3 + 1.5 * iqr, 3))
print("|média - mediana|:", round(abs(s.mean() - s.median()), 3))

n_out_inf = (s < (q1 - 1.5 * iqr)).sum()
n_out_sup = (s > (q3 + 1.5 * iqr)).sum()
print(f"outliers abaixo: {n_out_inf} | acima: {n_out_sup}")
```

**Insight (3.2):** NPS médio ~**4,45** e mediana ~**4,5** (bem próximas) — distribuição centrada em nota baixa, sem assimetria absurda. Desvio ~**2,47**; **0 outliers** pelo critério 1,5×IQR — as notas extremas parecem experiências reais, não erro de dado.

### 3.4 — `groupby`: NPS e % detratores por fator operacional

Cortes de operação: região, atraso, SAC, tentativas de entrega e reclamações.

```python
def resumo_nps_por(coluna):
    return (
        df_mov_ecom_filtrado.groupby(coluna, observed=True)[["nps_score", "eh_detrator"]]
        .agg(
            n=("nps_score", "size"),
            nps_medio=("nps_score", "mean"),
            nps_mediano=("nps_score", "median"),
            pct_detratores=("eh_detrator", "mean"),
        )
        .round(3)
    )


print("=== Região ===")
display(resumo_nps_por("customer_region"))

print("=== Faixa de atraso ===")
display(resumo_nps_por("faixa_atraso"))

print("=== Faixa SAC ===")
display(resumo_nps_por("faixa_sac"))

print("=== Tentativas de entrega ===")
display(resumo_nps_por("delivery_attempts"))

print("=== Faixa de reclamações ===")
display(resumo_nps_por("faixa_reclamacoes"))
```

**Insight (3.4):** fatores **operacionais** saltam aos olhos:
- **Atraso:** NPS cai de ~6,7 (0 dias) para ~1,3 (5+ dias); % detratores sobe de ~39% para ~99%.
- **SAC:** mais contatos → pior NPS (0 contatos ~5,5 vs 4+ ~2,5).
- **Reclamações:** 0–2 reclamações NPS ~6,6 vs 6+ ~2,9.
- **Região / tentativas de entrega:** variação modesta — operação pesa mais que geografia.

```python
print("=== Faixa de idade ===")
display(resumo_nps_por("faixa_idade"))

print("=== Faixa de tenure ===")
display(resumo_nps_por("faixa_tenure"))

print("=== Faixa de ticket (quartis de order_value) ===")
display(resumo_nps_por("faixa_ticket"))
```

**Insight (3.5):** perfil de cliente explica **menos** que operação:

- **Tenure:** clientes até 12 meses um pouco piores (~4,17 vs ~4,4–4,7) — possível sinal de onboarding/experiência inicial, mas efeito menor que atraso/SAC.

## Parte 5 — Registrar hipóteses (antes da estatística)

**Objetivo:** transformar achados das Partes 3 e 4 em **hipóteses testáveis**. Hipótese da EDA **não é prova** — a Parte 6 entra com IC, testes e regressão.

**Framework da aula:** Observação → Evidência (número/gráfico) → Hipótese → Próximos passos.

Base: `df_mov_ecom_filtrado`.

### 5.1 — Extrair evidências numéricas (recapitulando Partes 3–4)

Consolidamos os números que sustentam cada hipótese — para não repetir achados "de memória".

```python
def evidencia_sac():
    g = (
        df_mov_ecom_filtrado.groupby("faixa_sac", observed=True)
        .agg(n=("nps_score", "size"), nps_medio=("nps_score", "mean"), pct_detratores=("eh_detrator", "mean"))
        .round(3)
    )
    corr_sac = df_mov_ecom_filtrado[["customer_service_contacts", "csat_internal_score", "nps_score"]].corr().round(3)
    return g, corr_sac


def evidencia_tentativas():
    return (
        df_mov_ecom_filtrado.groupby("delivery_attempts", observed=True)
        .agg(n=("nps_score", "size"), nps_medio=("nps_score", "mean"), pct_detratores=("eh_detrator", "mean"))
        .round(3)
    )


def evidencia_atraso():
    g = (
        df_mov_ecom_filtrado.groupby("faixa_atraso", observed=True)
        .agg(
            n=("nps_score", "size"),
            nps_medio=("nps_score", "mean"),
            nps_mediano=("nps_score", "median"),
            pct_detratores=("eh_detrator", "mean"),
        )
        .round(3)
    )
    r = df_mov_ecom_filtrado[["delivery_delay_days", "nps_score"]].corr().iloc[0, 1].round(3)
    return g, r


def evidencia_reclamacoes():
    return (
        df_mov_ecom_filtrado.groupby("faixa_reclamacoes", observed=True)
        .agg(n=("nps_score", "size"), nps_medio=("nps_score", "mean"), pct_detratores=("eh_detrator", "mean"))
        .round(3)
    )


print("=== H1 — SAC / CSAT ===")
display(evidencia_sac()[0])
display(evidencia_sac()[1])

print("=== H2 — Tentativas de entrega ===")
display(evidencia_tentativas())

print("=== H3 — Atraso ===")
g_atraso, r_atraso = evidencia_atraso()
display(g_atraso)
print(f"Correlação Pearson atraso × NPS: {r_atraso}")

print("=== H4 — Reclamações ===")
display(evidencia_reclamacoes())
```

**Insight (5.1):** este bloco amarra as Partes 3–4 em números reutilizáveis. Repare no contraste de magnitude:

- **SAC** acompanha essa lógica (0 contatos → NPS ~5,5; 4+ → ~2,5).
- **Tentativas de entrega** oscilam pouco (~4,3–4,6) — hipótese mais fraca, mas ainda testável na Parte 6.

Isso orienta **prioridade**: H3 e H4 primeiro; H1 em seguida; H2 com expectativa moderada.

|---|------------|------------------------|----------|---------------------------|
| **H1** | Quanto mais contatos no SAC, pior o NPS; CSAT interno alto quando há contato | SAC 0 → NPS ~5,5 / 60% detratores; 4+ → ~2,5 / 94%. Corr. SAC×NPS ≈ −0,35; CSAT×NPS ≈ +0,55 (proxy/leakage) | Cliente aciona o SAC quando já há problema; **mais contatos ↔ pior NPS**. A experiência **quebra de verdade a partir de 3 dias de atraso** (NPS mediano cai de ~5 para ~3) ou quando o cliente precisa **ligar 2+ vezes** no SAC. Perfil — região, idade, ticket — **quase não separa** quem gosta ou não; clientes **novos (até 12 meses)** só ficam um pouco piores. **Próximo passo operacional:** (1) alerta e ação no **2º dia de atraso**, antes do degrau de 3 dias; (2) resolver no **1º contato SAC**; (3) tratar **reclamações a partir da 3ª** como incidente crítico; (4) medir NPS por **pedido no prazo vs. atrasado**, não por região.

**Insight (7.5 — pitch):** este texto cabe numa reunião de **30 segundos** porque segue a ordem que o gerente pensa: (1) qual o problema, (2) quão grave, (3) onde quebra, (4) quem afeta, (5) o que fazer segunda-feira. Os quatro passos operacionais são **preventivos** — atuam antes da pesquisa NPS, alinhados ao objetivo do case (antecipar insatisfação). Evite prometer “melhorar SAC” isolado: os números mostram SAC como **efeito**, não causa raiz.

### 7.6 — Ressalvas, limitações e próximos passos

**O que deixar claro na conversa com o gerente:**

1. **Amostra, não universo:** analisamos **2.356 pedidos** (excluímos ~144 inconsistentes — ver veredito 2.6). Os percentuais descrevem **esta base**, não necessariamente toda a operação ao vivo.

2. **Correlação ≠ causa:** atraso, SAC e reclamações **andam juntos**. Melhorar entrega provavelmente reduz SAC e reclamações — mas isso ainda será testado formalmente na **Parte 6** (IC, testes, regressão).

3. **NPS transacional:** 1 linha = 1 **pedido**. O mesmo cliente em outro pedido pode ter outra nota.

4. **Variáveis com cuidado:** `csat_internal_score` e `repeat_purchase_30d` têm risco de **leakage** (medem depois ou consequência da experiência) — não entraram nas respostas ao gerente como alavancas de ação.

5. **O que ainda falta:** Parte 6 confirma se as diferenças (ex.: 3 vs. 0 dias de atraso) são estatisticamente robustas e quantifica o efeito isolado de cada fator.
